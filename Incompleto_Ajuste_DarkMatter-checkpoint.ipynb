```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import glob

# Constante gravitacional (kpc/solar mass*(km/s)^2)
G = 4.302e-6

def disco(R, Md, d):
    """ Modelo de velocidad de rotación debido al disco galáctico. """
    return np.sqrt(G * Md / (R**2 + d**2)**0.75)

def halo(R, rho0, rs):
    """ Modelo de velocidad de rotación debido al halo de materia oscura (perfil NFW). """
    x = R / rs
    M = 4 * np.pi * rho0 * rs**3 * (np.log(1 + x) - x / (1 + x))
    return np.sqrt(G * M / R)

def ajustar_y_demostrar():
    # Preparación del gráfico
    fig, axs = plt.subplots(2, 1, figsize=(12, 12), sharex=True)

    for archivo in sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat")):
        R, V = np.loadtxt(archivo, skiprows=13, unpack=True)

        if R.max() > 10:  # Aseguramos datos suficientes para un buen ajuste
            # Ajuste solo con el disco
            #INCOMPLETO!!!
            popt_disco, _ = ZZZ
            V_fit_disco = disco(R, *popt_disco)

            # Ajuste con disco y halo
            def modelo_completo(R, Md, d, rho0, rs):
                return np.sqrt(disco(R, Md, d)**2 + halo(R, rho0, rs)**2)

            #INCOMPLETO!!!
            popt_total, _ = ZZZ
            V_fit_total = modelo_completo(R, *popt_total)

            # Visualización de los resultados
            axs[0].plot(R, V, label='Datos')
            axs[0].plot(R, V_fit_disco, label='Ajuste solo con disco')
            axs[0].set_title('Ajuste utilizando solo el componente de disco')
            axs[0].set_xlabel('Radio (kpc)')
            axs[0].set_ylabel('Velocidad de rotación (km/s)')

            axs[1].plot(R, V, label='Datos')
            axs[1].plot(R, V_fit_total, label='Ajuste con disco y halo')
            axs[1].set_title('Ajuste utilizando el componente de disco y halo de materia oscura')
            axs[1].set_xlabel('Radio (kpc)')
            axs[1].set_ylabel('Velocidad de rotación (km/s)')

    for ax in axs:
        ax.legend()
        ax.grid(True)

    plt.tight_layout()
    plt.show()

ajustar_y_demostrar()
```

Se ve claramente cómo el modelo sin materia oscura falla en ajustar las curvas de rotación de galaxias.

Solo si se modifica la gravedad como en teorías de MOND se alacanzan ajustes similares sin introducir la materia oscura. El problema de teorías como MOND es que es totalmente arbitraria la escala a la que se modifica la ley de gravedad y la dependencia con la distancia sin venir de principios fundamentales, como si lo hace la materia oscura.

```python

def chi_squared_reduced(observado, modelado, error, num_params):
    return np.sum(((observado - modelado) / error) ** 2) / (len(observado) - num_params)

def coef_determinacion(observado, modelado):
    ss_res = np.sum((observado - modelado) ** 2)
    ss_tot = np.sum((observado - np.mean(observado)) ** 2)
    return 1 - ss_res / ss_tot
```

```python
# Ajustar y evaluar bondad de ajuste
def ajustar_y_evaluar():
    fig, axs = plt.subplots(2, 1, figsize=(10, 10), sharex=True)

    resultados = []

    for archivo in sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat"))[:5]:  # Limitamos a 5 galaxias para simplificar
        R, V = np.loadtxt(archivo, skiprows=13, unpack=True)
        error = np.full_like(V, 5)  # Asumimos un error constante si no se especifica

        # Modelo sólo con disco
        #INCOMPLETO!!!
        popt_disco, _ = ZZZ
        V_fit_disco = disco(R, *popt_disco)
        chi2_disco = chi_squared_reduced(V, V_fit_disco, error, len(popt_disco))
        r2_disco = coef_determinacion(V, V_fit_disco)

        # Modelo con disco y halo
        def modelo_completo(R, Md, d, rho0, rs):
            return np.sqrt(disco(R, Md, d)**2 + halo(R, rho0, rs)**2)

        #INCOMPLETO!!!
        popt_total, _ = ZZZ
        V_fit_total = modelo_completo(R, *popt_total)
        chi2_total = chi_squared_reduced(V, V_fit_total, error, len(popt_total))
        r2_total = coef_determinacion(V, V_fit_total)

        resultados.append((archivo.split('/')[-1], chi2_disco, r2_disco, chi2_total, r2_total))

        # Ploteo de resultados
        ax = axs[0 if archivo == sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat"))[0] else 1]
        ax.plot(R, V, 'k:', label='Datos' if archivo == sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat"))[0] else "")
        ax.plot(R, V_fit_disco, 'r--', label='Solo disco' if archivo == sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat"))[0] else "")
        ax.plot(R, V_fit_total, 'b-', label='Disco + Halo' if archivo == sorted(glob.glob("/content/drive/MyDrive/AJUSTES/rotation_curves/*.dat"))[0] else "")

    axs[0].legend()
    axs[0].set_ylabel('Velocidad de rotación (km/s)')
    axs[1].set_xlabel('Radio (kpc)')
    axs[1].set_ylabel('Velocidad de rotación (km/s)')
    plt.show()

    return resultados

resultados = ajustar_y_evaluar()
for res in resultados:
    print(f"Galaxia: {res[0]}, Chi2 Disco: {res[1]:.2f}, R2 Disco: {res[2]:.2f}, Chi2 Total: {res[3]:.2f}, R2 Total: {res[4]:.2f}")
```
